```python
def check_password(password):
    length = False
    digit = False
    uppercase = False
    if len(password)>=8:
        length = True
    for i in password:
        if i.isdigit():
            digit = True
        elif i.isupper():
            uppercase = True
    if length and digit and uppercase:
        return True
    else:
        return False
print(check_password("manju7"))
print(check_password("Manju1234"))
```

```python
class Employee: # parent
    def __init__(self, name, salary):
         self.name = name
         self.salary = salary

    def get_info(self):
      return f'{self.name} earns {self.salary}'

class Manager(Employee):
     def __init__(self,name,salary, teamsize):
          super().__init__(name,salary)
          self.teamsize = teamsize

     def complete_info(self):
          basic_info = self.get_info()  # extending the parent method instead of repeating the codes
          return f'{basic_info}, he manages a team of size {self.teamsize}'


emp = Manager("arun", 80000, 10)
emp.complete_info()
```
