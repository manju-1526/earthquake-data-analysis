```python
import requests
import pandas as pd
from datetime import datetime
```

```python
url = "https://earthquake.usgs.gov/fdsnws/event/1/query"
```

```python
all_records = []
start_year = datetime.now().year - 5 # last 5 years
end_year = datetime.now().year
```

```python
for year in range(start_year, end_year + 1):
    for month in range(1, 13):

        start_date = f"{year}-{month:02d}-01"
        if month == 12:
            end_date = f"{year+1}-01-01"
        else:
            end_date = f"{year}-{month+1:02d}-0"
params = {
    "format": "geojson",
    "starttime": start_date,
    "endtime": end_date,
    "minmagnitude": 3
}
```

```python
response = requests.get(url, params=params)

if response.status_code != 200:
    print("Request failed:", response.text[:200])
else:
    print("Success!")
print(response)
```

```python
data = response.json()
for f in data["features"]:
     p = f["properties"] 
     g = f["geometry"]["coordinates"]
     all_records.append({
          "id": f.get("id"),
          "time": pd.to_datetime(p.get("time")),
          "updated": pd.to_datetime(p.get("updated")),
          "latitude": g[1] if g else none,
          "longitude": g[0] if g else none,
          "depth_km": g[2] if g else none,
          "mag": p.get("mag"),
          "magType": p.get("magType"),
          "place": p.get("place"),
          "status": p.get("status"),
          "tsunami": p.get("tsunami"),
          "alert": p.get("alert"),
          "felt": p.get("felt"),
          "cdi": p.get("cdi"),
          "mmi": p.get("mmi"),
          "sig": p.get("sig"),
          "net": p.get("net"),
          "code": p.get("code"),
          "ids": p.get("ids"),
          "sources": p.get("sources"),
          "types": p.get("types"),
          "nst": p.get("nst"),
          "dmin": p.get("dmin"),
          "rms": p.get("rms"),
          "gap": p.get("gap"),
          "type": p.get("type")
     })
     df = pd.DataFrame(all_records)
```
